## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

This project helps content editors decide exactly which 10 to 20 articles they should update in their next work cycle. If the model makes a wrong call, editors either waste 5 to 10 hours fixing a dead page (false positive) or lose revenue because a major declining page was missed (false negative). A simple rule like "update any page down 20%" isn't enough because search rankings depend on a tricky combination of article type, length, search position, and user reading behavior

```python
import pandas as pd

# Load dataset
df = pd.read_csv("/content/content_refresh_anonymized.csv")

# Find the label column dynamically to avoid KeyError
label_col = next(
    (col for col in df.columns if "declining" in col.lower()), None
)

print("=== PART 2: BASE RATE & DECISION TARGET ===")
if label_col:
  target_counts = df[label_col].value_counts(dropna=False)
  target_percentages = df[label_col].value_counts(
      dropna=False, normalize=True
  )

  print(f"Target Column Detected: '{label_col}'")
  for label, count in target_counts.items():
    pct = target_percentages[label] * 100
    print(f" - Status [{label}]: {count:,} items ({pct:.2f}%)")
else:
  print("Available columns in dataset:")
  print(df.columns.tolist())
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Checking the 30,000-row starter dataset reveals that 1,205 rows (4.02%) have an avg_position of 0, which means "no rank data available" rather than rank zero. Additionally, rate metrics like Click-Through Rate (CTR) are stored as percentages with an overall mean of 0.76%, and we must strictly remove the trend_pct and trend_direction columns from our features so the model doesn't cheat using the target label.

```python
import pandas as pd

# Load dataset
df = pd.read_csv("/content/content_refresh_anonymized.csv")

# Part 3 Check: Key dataset statistics & gotchas
total_rows = len(df)
unranked_count = (df["avg_position"] == 0).sum()
unranked_pct = (unranked_count / total_rows) * 100
mean_ctr = df["ctr"].mean()

print("=== PART 3: DATA GOTCHAS & REAL NUMBERS ===")
print(
    f"1. Unranked Pages (avg_position == 0) : {unranked_count:,} ({unranked_pct:.2f}%)"
)
print(f"2. Mean CTR (Stored as x100 scale)   : {mean_ctr:.2f}%")
print("\n3. Content Type Breakdown:")
print(df["content_type"].value_counts().to_string())
```
