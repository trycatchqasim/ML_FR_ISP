## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Checking the 30,000-row starter dataset reveals that 1,205 rows (4.02%) have an avg_position of 0, which means "no rank data available" rather than rank zero. Additionally, rate metrics like Click-Through Rate (CTR) are stored as percentages with an overall mean of 0.76%, and we must strictly remove the trend_pct and trend_direction columns from our features so the model doesn't cheat using the target label.

```python

# 1. Unranked position check (avg_position == 0 means missing data)
unranked_count = (df["avg_position"] == 0).sum()
unranked_pct = (unranked_count / total_rows) * 100

# 2. Rate column check (ctr is stored x100)
mean_ctr = df["ctr"].mean()

# 3. Label / Target check
target_counts = df["is_declining_label"].value_counts(dropna=False).to_dict()

print("\n[Part 3] Real Numbers from Data:")
print(
    f" - Unranked items (avg_position == 0) : {unranked_count:,} ({unranked_pct:.2f}%)"
)
print(f" - Mean CTR (stored x100)           : {mean_ctr:.2f}%")
print(f" - Declining Label Distribution     : {target_counts}")
```
